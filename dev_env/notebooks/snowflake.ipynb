```python
import pandas as pd
import snowflake.connector
from snowflake.connector.pandas_tools import write_pandas

account   = "ubazxip-fcb79770"
user      = "DERIVZERO"
role      = "ACCOUNTADMIN"
warehouse = "APP_WH"
database  = "FORECAST_DB"
schema    = "APP"

conn = snowflake.connector.connect(
    account=account,
    user=user,
    authenticator="externalbrowser",
    role=role,
    warehouse=warehouse,
    database=database,
    schema=schema,
)

df = pd.DataFrame({
    "id": [1, 2, 3, 4, 5],
    "value": ["a", "b", "c", "d", "e"]
})

success, nchunks, nrows, _ = write_pandas(
    conn,
    df,
    "SMOKE_TEST",
    auto_create_table=True
)

print("success:", success, "rows:", nrows)
conn.close()
```
